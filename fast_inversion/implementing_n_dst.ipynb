```python
def print_matrix_rounded(L, precision=3):
    # Round the entries
    L_rounded = np.round(L, precision)
    # Print each row
    for row in L_rounded:
        print("[ " + "  ".join(f"{val:.{precision}f}" for val in row) + " ]")
```

```python
def make_T(n):
    T = np.zeros((n, n), dtype=complex)
    T[0, 0] = 1
    T[n//2, n//2] = 1
    for i in range(1, n//2):
        T[i, i] = 1/np.sqrt(2)
        T[i, n//2+i] = complex(0, (1/np.sqrt(2)))
    for i in range(n//2 + 1, n):
        T[i, n-i] = 1/np.sqrt(2)
        T[i, n-i+n//2] = complex(0, -(1/np.sqrt(2)))
    
    return T
        
   
```

```python
def make_T2(n):
    T = np.zeros((n, n), dtype=complex)
    T[0, 0] = 1
    T[n//2-1, n//2-1] = 1
    for i in range(1, n//2-1):
        T[i, i] = 1/np.sqrt(2)
        T[i, n-i] = complex(0, (1/np.sqrt(2)))
    for i in range(n//2, n):
        T[i, n-i-1] = 1/np.sqrt(2)
        T[i, n-i+n//2-1] = complex(0, -(1/np.sqrt(2)))
    
    return T
```

```python
import numpy as np

def make_T_corrected(n):
    """
    n must be even, n = 2*N.
    Returns the unitary T such that T† F_{2N} T = C_N ⊕ S_N.
    """
    assert n % 2 == 0, "n must be even"
    N = n // 2
    coef = 1/np.sqrt(2)

    T = np.zeros((n, n), dtype=complex)
    # fixed points
    T[0, 0] = 1
    T[N, N] = 1

    # 1 ≤ k ≤ N-1
    for k in range(1, N):
        T[k,     k] = coef          # real part on the diagonal
        T[n-k,   k] =  1j * coef    # +i in row n−k, col k

    # N+1 ≤ k ≤ 2N−1
    for k in range(N+1, n):
        T[k,     k] = coef
        T[n-k,   k] = -1j * coef    # –i in row n−k, col k

    return T
```

```python
def make_V(n):
    N = 2*n 
    V = np.zeros((N, N), dtype=complex)
    
    for i in range(n):
        V[i, i] = 1
        V[i, n+i] = 1
        V[n+i, n-i-1] = 1
        V[n+i, 2*n-i-1] = -1

    return V
```

```python
# Use permutation matrices to adjust the V_N matrix so that the top left corner is the V_M matrix (where M<N and N=2^n)

N = 8 # (half of) the size of V_N matrix created with standard method from Klappenecker paper (N=2^n), matrix will be (2N x 2N)
M = 6 # (half of) the size of adjusted V_N matrix (M != 2^n), matrix will be (2M x 2M)
V_N = make_V(N)

############################### Make P_row ###############################
P_row = np.zeros((2*N, 2*N), dtype=complex)
# set ones in P_row (short version) (map from column to row a.k.a. input to output)
for c in range(2*N): # c is the column number
    if c < M:
        P_row[c,c] = 1
    elif c >= M and c < 2*N-M:
        P_row[c+M,c] = 1
    elif c >= 2*N-M and c < 2*N:
        P_row[c + 2*M - 2*N,c] = 1
#print("P_row: ")
#print_matrix_rounded(P_row, precision=2)


############################### Make P_col ###############################
P_col = np.zeros((2*N, 2*N), dtype=complex)
# set ones in P_col (short version) (map from column to row a.k.a. input to output)
for c in range(2*N): # c is the column number
    if c < M:
        P_col[c,c] = 1
    elif c == M:
        P_col[c+(N-M),c] = 1
    elif c >= M+1 and c < 2*M:
        P_col[c + N - M,c] = 1
    elif c >= 2*M and c < N+M:
        P_col[c-M,c] = 1
    elif c >= N+M and c < 2*N:
        P_col[c,c] = 1
#print("P_col: ")
#print_matrix_rounded(P_col, precision=2)

print("P_col is unitary? ", np.allclose(np.eye(2*N),P_col @ np.conj(P_col.T)))
print("P_row is unitary? ", np.allclose(np.eye(2*N),P_row @ np.conj(P_row.T)))

############################### Apply P_row and P_col to V_N to get V_M ###############################
V_M = P_row @ V_N @ P_col
print("V_M: ")
print_matrix_rounded(V_M, precision=2)

V_M_actual = make_V(M)
print("Is the upper (2M x 2M) corner of the permuted V_N equal to V_M: ", str(np.allclose(V_M_actual, V_M[:2*M,:2*M])))
```

```python
def make_U(n):

    N = n*2
    U = np.zeros((N, N), dtype=complex)

    
    for i in range(0, n):
        omega = np.exp(2*np.pi*1j/(4*n))
        U[i, i] = omega.conj()**(i)/np.sqrt(2) 
        U[i, n+i-1] = - 1j * omega.conj()**i/np.sqrt(2)

    for i in range(1, n):
        omega = np.exp(2*np.pi*1j/(4*n))
        U[n+i, n-i] = omega ** (n-i)/np.sqrt(2) 
        U[n+i, 2*n-i-1] = -1j * omega ** (n-i)/np.sqrt(2)


    U[0, 0] = 1
    U[0, n-1] = 0
    U[n, 2*n-1] = -1

    return U
    
```

```python
# Use permutation matrices to adjust the U_N matrix so that the top left corner is the U_M matrix (where M<N and N=2^n)
import math

N = 16 # (half of) the size of U_N matrix created with standard method from Klappenecker paper (N=2^n), matrix will be (2N x 2N)
M = 14 # (half of) the size of adjusted U_N matrix (M != 2^n), matrix will be (2M x 2M)
U_N = make_U(N)

############################### Make P_row ###############################
P_row = np.zeros((2*N, 2*N), dtype=complex)
# set ones in P_row (short version) (map from column to row a.k.a. input to output)
for c in range(2*N): # c is the column number
    if c < M:
        P_row[c,c] = 1
    elif c >= M and c < N:
        P_row[c+M,c] = 1
    elif c==N:
        P_row[c+M-N,c] = 1
    elif c >= N+1 and c < 2*N-M+1:
        P_row[c+M-1,c] = 1
    elif c >= 2*N-M+1 and c < 2*N:
        P_row[c + 2*M - 2*N,c] = 1
#print("P_row: ")
#print_matrix_rounded(P_row, precision=2)


############################### Make P_col ###############################
P_col = np.zeros((2*N, 2*N), dtype=complex)
# set ones in P_col (short version) (map from column to row a.k.a. input to output)
for c in range(2*N): # c is the column number
    if c < M:
        P_col[c,c] = 1
    elif c >= M and c < 2*M-1:
        P_col[c + N - M,c] = 1
    elif c == 2*M-1:
        P_col[c+2*(N-M),c] = 1
    elif c >= 2*M and c < N+M:
        P_col[c-M,c] = 1
    elif c >= N+M and c < 2*N:
        P_col[c-1,c] = 1
#print("P_col: ")
#print_matrix_rounded(P_col, precision=2)

############################### Make Gamma ###############################

# The Gamma matrix is just the kronecker product of log(N) single gate rotation matrices (Phase gate rotation about the Z axis)
n = math.floor(math.log2(N))
l = 1
gamma_m_n = np.exp(2j * math.pi * (1/M - 1/N) / 4)
Gamma = np.array([[1, 0],[0,1/gamma_m_n]])
for i in range(n-1):
    l *= 2
    mat = np.array([[1, 0],[0,np.power(gamma_m_n,-l)]])
    Gamma = np.kron(mat, Gamma)
Gamma = np.kron(np.array([[1, 0],[0,np.power(gamma_m_n,2*l-1)]]), Gamma)

############################### Make P_gamma_row ###############################
P_gamma_row = np.zeros((2*N, 2*N), dtype=complex)
# set ones in P_gamma_row (short version) (map from column to row a.k.a. input to output)
for c in range(2*N): # c is the column number
    if c < M:
        P_gamma_row[c,c] = 1
    elif c >= M and c < 2*N-M:
        P_gamma_row[c+M,c] = 1
    elif c >= 2*N-M and c < 2*N-1:
        P_gamma_row[c+2*(M-N)+1,c] = 1
    elif c == 2*N-1:
        P_gamma_row[c-2*N+1+M,c] = 1

############################### Make P_gamma_col ###############################
P_gamma_col = np.zeros((2*N, 2*N), dtype=complex)
# set ones in P_gamma_col (short version) (map from column to row a.k.a. input to output)
for c in range(2*N): # c is the column number
    if c < M:
        P_gamma_col[c,c] = 1
    elif c == M:
        P_gamma_col[c+2*N-M-1,c] = 1
    elif c >= M+1 and c < 2*M:
        P_gamma_col[c+2*(N-M)-1,c] = 1
    elif c >= 2*M and c < 2*N:
        P_gamma_col[c-M,c] = 1

#print_matrix_rounded(P_gamma_row @ Gamma @ P_gamma_col, precision=2)

print("P_col is unitary? ", np.allclose(np.eye(2*N),P_col @ np.conj(P_col.T)))
print("P_row is unitary? ", np.allclose(np.eye(2*N),P_row @ np.conj(P_row.T)))
print("Gamma is unitary? ", np.allclose(np.eye(2*N),Gamma @ np.conj(Gamma.T)))
print("P_gamma_col is unitary? ", np.allclose(np.eye(2*N),P_gamma_col @ np.conj(P_gamma_col.T)))
print("P_gamma_row is unitary? ", np.allclose(np.eye(2*N),P_gamma_row @ np.conj(P_gamma_row.T)))


############################### Apply P_row and P_col to T_N_prime to get T_M ###############################
U_M = (P_gamma_row @ Gamma @ P_gamma_col) @ P_row @ U_N @ P_col
print("U_M created: ")
print_matrix_rounded(U_M, precision=2)

U_M_actual = make_U(M)
print("U_M actual: ")
print_matrix_rounded(U_M_actual, precision=2)
print("Is the upper (2M x 2M) corner of the permuted U_N equal to U_M: ", str(np.allclose(U_M_actual, U_M[:2*M,:2*M])))
```
